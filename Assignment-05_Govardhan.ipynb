### Problem-1

In a network, data packets are sent through a shared channel. Suppose there are 10 devices attempting to send data. In any given time slot, each device has a 20% chance of attempting to send a packet, independently of other devices. A collision occurs if two or more devices attempt to send a packet in the same time slot. What is the probability that in a given time slot, there will be no collisions?

### Solution 1

Let $X$ be the number of devices attempting to send data in a given time slot.
Each Device independent attempts to send with the probability $p = 0.2$
There are $n = 10$ Devices,

Thus,

$$
X \sim \text{Binomial}(n=10,p=0.2)
$$

The Probability mass function of a Binomial Distribution is:

$$
P(X=K) = \binom{n}{k} p^k (1-p)^{n-k}
$$ 

A collision occurs if two or more devices transmit simultaneously.
Therefore **no collision** occurs if:

$$
X = 0 \quad \text{or} \quad X=1
$$

Hence,

$$
P(\text{No Collision}) = P(X=0) + P(X=1)
$$

### Step1: Compute $P(X=0)$

$$
P(X=0) = \binom{10}{0} (0.2)^0 (0.8)^{10}
$$

$$
P(X=0) = (0.8)^{10}
$$

$$
 = 0.1074
$$



### Step 2 Compute $P(X=1)$

$$
P(X=1) = \binom{10}{1} (0.2)^1 (0.8)^9
$$

$$
 = 10 \times 0.2 \times (0.8)^9
$$

$$
= 0.2684
$$


### Step 3: Final Probability

$$
P(\text{No Collision}) = 0.1074 + 0.2684
$$

$$
= 0.3758
$$


### Final Answer 

$$
\boxed{P(\text{No Collision}) \approx 0.376}
$$

## Additional Knowledge: Binomial Distribution

The Binomial Distribution models the number of successes in $n$ independent 
Bernoulli trials, where each trial has only two outcomes: success or failure, 
with constant probability $p$ of success.

Formally, if $X$ represents the number of successes, then:

$$
X \sim \text{Bin}(n, p)
$$

The probability mass function (PMF) is:

$$
P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}, \quad 0 \le k \le n
$$

where:
- $n$ = total number of trials,
- $p$ = probability of success in each trial,
- $\binom{n}{k}$ = number of ways to choose $k$ successes from $n$ trials.

The mean and variance of the binomial distribution are:

$$
E[X] = np
$$

$$
\text{Var}(X) = np(1-p)
$$

The distribution originates from the Binomial Theorem and was formalized by 
Jacob Bernoulli in the 17th century while studying repeated random experiments. 
It is widely used in quality control, medical testing, communication networks, 
and machine learning for modeling binary outcomes.

### Problem-3

A software project has two modules, A and B. The probability that module A has a bug is 0.15, and the probability that module B has a bug is 0.10. The probability that both modules have bugs is 0.03.

(a) What is the probability that module A has a bug, given that module B has a bug?

(b) What is the probability that module B has a bug, given that module A does not have a bug?

### Solution

Given

$$
P(A) = 0.15, \quad P(B) 0.10 \quad P(A \cap B) = 0.03
$$


### (a) Probability that module A has a bug given B has a bug

Using conditional probability formula:

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

$$
= \frac{0.03}{0.10}
$$

$$
= 0.30
$$

Thus,

$$
\boxed{P(A \mid B) = 0.30}
$$

### (b) Probability that module B has a bug given A does not have a bug

First, compute:

$$
P(A^c) = 1 - P(A) = 1 - 0.15 = 0.85
$$

Next,

$$
P(B \cap A^c) = P(B) - P(A \cap B)
$$

$$
= 0.10 - 0.03 = 0.07
$$

Now apply conditional probability

$$
P(B \mid A^c) = \frac{P(B \cap A^c)}{P(A^c)}
$$

$$
= \frac{0.07}{0.85}
$$

$$
\approx 0.082
$$

Thus,

$$
\boxed{P(B \mid A^c) \approx 0.082}
$$

The probability density function (pdf) of the triangular distribution is characterized by three points as shown in the figure below.

<img src="https://upload.wikimedia.org/wikipedia/commons/4/45/Triangular_distribution_PMF.png" alt="triangular distribution" width="500"/>

(a) min value of the distribution,
(b) max value of the distribution, and
(c) mode of the distribution

Note: $a \leq c \leq b$

Create a class `TriangularDistribution` with attributes `min`, `max` and `mode`. You need to implement the following methods:

- `pdf(x)`: returns the value of the pdf at point $x$.
- `plot_pdf()`: plots the pdf in the range $[min , max]$
- `cdf(x)`: returns the value of the cdf at point $x$ ; which denotes $P(X \leq x)$
- `plot_cdf()`: plots the cdf in the range $[min , max]$
- `get_probability(x1, x2)`: returns the probability $P(x_1 \leq X \leq x_2)$, where $x_1 \lt x_2$.

Gracefully handle the possible errors and also the edge conditions when either $ a = c $ or $b = c$ in the distribution.

Test your implementation.

````python
### Write your code here

import numpy as np
import matplotlib.pyplot as plt


class TriangularDistribution:
    
    def __init__(self, min_val, max_val, mode):
        if min_val > max_val:
            raise ValueError("min must be <= max")
        if not (min_val <= mode <= max_val):
            raise ValueError("mode must satisfy min <= mode <= max")
        
        self.a = min_val
        self.b = max_val
        self.c = mode

    
    def pdf(self, x):
        a, b, c = self.a, self.b, self.c
        
        if x < a or x > b:
            return 0.0
        
        if a == b:
            raise ValueError("Invalid distribution: min and max cannot be equal")
        
       
        if a <= x <= c:
            if c == a:  
                return 2 * (b - x) / ((b - a) * (b - c))
            return 2 * (x - a) / ((b - a) * (c - a))
        
      
        if c < x <= b:
            if c == b:  
                return 2 * (x - a) / ((b - a) * (c - a))
            return 2 * (b - x) / ((b - a) * (b - c))

    
    def cdf(self, x):
        a, b, c = self.a, self.b, self.c
        
        if x <= a:
            return 0.0
        if x >= b:
            return 1.0
        
        if a <= x <= c:
            if c == a:
                return 1 - ((b - x)**2) / ((b - a) * (b - c))
            return ((x - a)**2) / ((b - a) * (c - a))
        
        if c < x <= b:
            if c == b:
                return ((x - a)**2) / ((b - a) * (c - a))
            return 1 - ((b - x)**2) / ((b - a) * (b - c))

    
    def get_probability(self, x1, x2):
        if x1 >= x2:
            raise ValueError("x1 must be less than x2")
        
        return self.cdf(x2) - self.cdf(x1)

   
    def plot_pdf(self):
        x = np.linspace(self.a, self.b, 1000)
        y = [self.pdf(val) for val in x]
        
        plt.figure()
        plt.plot(x, y)
        plt.title("Triangular Distribution PDF")
        plt.xlabel("x")
        plt.ylabel("f(x)")
        plt.show()

 
    def plot_cdf(self):
        x = np.linspace(self.a, self.b, 1000)
        y = [self.cdf(val) for val in x]
        
        plt.figure()
        plt.plot(x, y)
        plt.title("Triangular Distribution CDF")
        plt.xlabel("x")
        plt.ylabel("F(x)")
        plt.show()
````

### Problem-14

A grayscale image can be represented as a matrix where the dimensions are (height, width). 
Let's consider a simplified scenario with two grayscale images, $A$ and $B$, both represented as 10×10 matrices where each element represents pixel intensity. Pixel intensity is an integer in the range [0-255].

You want to create a blended image $C$ by combining $A$ and $B$ such that $C=w_1 A+w_2 B$. This represents a weighted average blend.
Note that each element in the resultant matrix $C$ should also fall within the range [0, 255]. 

write the code of the function `blend_images(image_A, image_B, weight_A, weight_B)` that takes two matrices and two scalar weights and produces another matrix as an output by performing the above operation.

````python
import numpy as np

def blend_images(image_A, image_B, weight_A, weight_B):
    
    if image_A.shape != image_B.shape:
        raise ValueError("Images must have same dimensions")
    
    height, width = image_A.shape
    result = np.zeros((height, width), dtype=np.uint8)
    
    for i in range(height):
        for j in range(width):
            value = weight_A * image_A[i][j] + weight_B * image_B[i][j]
            
            # Clip manually
            if value < 0:
                value = 0
            elif value > 255:
                value = 255
            
            result[i][j] = int(value)
    
    return result


A = np.random.randint(0, 256, (10, 10), dtype=np.uint8)
B = np.random.randint(0, 256, (10, 10), dtype=np.uint8)


C = blend_images(A, B, 0.6, 0.4)

print(C)

````

Calculate the eigen values and eigen vectors of the following matrices.

$A = \begin{bmatrix}
5 & 2 \\
2 & 2
\end{bmatrix}$      

$B = \begin{bmatrix}
1 & 1 & 0\\
0 & 2 & 0\\
-1 & 1 & 3
\end{bmatrix}$

````python
import numpy as np

A = np.array([[5,2],[2,2]])
B = np.array([[1,1,0],[0,2,0],[-1,1,3]])
eigvals, eigvecs = np.linalg.eig(A)
eigvalsb,eigvecsb = np.linalg.eig(B)

print("Eigenvalues:", eigvals)
print("Eigenvectors:\n", eigvecs)

print("EigenvaluesB:", eigvalsb)
print("Eigenvectors:\n", eigvecsb)
````
