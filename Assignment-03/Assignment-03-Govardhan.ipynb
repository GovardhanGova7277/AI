#### 12. What States have the highest estimated percent change between the 2010 Census and the 2017 Population Estimate?

**NOTE:** This will take several lines of code, as it requires a recalculation of Percent Change.

```python
# Write your code here

state_pop = pop.groupby('State')[['2010Census','2017PopEstimate']].sum()

state_pop['Percent_Change'] = (
    (state_pop['2017PopEstimate'] - state_pop['2010Census']) / state_pop['2010Census'].replace(0,pd.NA)
) * 100

state_pop.sort_values(by='Percent_Change',ascending=False).head()
```
