## Problem: 2

Generate an array of length $3n$ , filled with the cyclic pattern 1, 2, 3. Where $n$ is a natural number.

```python
# Write your code 
n = int(input("Enter Natural Number(1,2,3,..): "))
length = 3 * n
arr = [1,2,3]
result = []
for i in range(n):
    result.extend(arr)

result_array = np.array(result)

if length == len(result_array):
    print(result_array)
```

## Problem: 3
Create an array of the first $n$ odd integers. $n$ is a natural number input from user.

```python
# Write your code here

n = int(input("Enter natural number (1,2,3.....): "))
result = []
for i in range(1,n+1):
    if i%2==1:
        result.append(i)

result_array = np.array(result)
print(result_array)
```

## Problem: 4

Create a 10 x 10 arrays of zeros and then "frame" it with a border of ones.

```python
# Write your code here
arr = np.zeros((10,10),dtype=int)
arr[0,:] = 1
arr[-1,:] = 1
arr[:,0] = 1
arr[:,-1] = 1
print(arr)
```

## Problem: 5

Create an 8 x 8 array with a checkerboard pattern of zeros and ones.

```python
# Write your code here
checkerboard = np.zeros((8,8),dtype = int)
for i in range(8):
    for j in range(8):
        if (i+j)%2 == 0:
            checkerboard[i,j] = 1
print(checkerboard)
```

## Problem: 6

Create an array $x$ with 100 points in the interval $[-1,1]$ (the points must be uniformly spaced). Then create another array $y$. 

Where $y = x^2 sin(x^2) + x$. 

Try to plot $y$ vs $x$ using matplotlib.

```python
# Write your code here
import matplotlib.pyplot as plt
x = np.linspace(-1,1,101)
y = (x ** 2) * (np.sin(x ** 2)) + x

plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('y')
plt.title("Plot of y = x^2 sin(x^2) + x")
plt.grid(True)
plt.show()
```

## Problem: 7 
Create a function which creates an $n×n$ array with $(i , j)$ -entry equal to $i+ j$

```python
# Write your code here
def create_sum_matrix(n):
    matrix = []
    for i in range(n):
        rows = []
        for j in range(n):
            rows.append(i+j)
        matrix.append(rows)
    return matrix
print(np.array(create_sum_matrix(5)))
```

# Problem: 8
Evaluate $cos$ and $sin$ on the interval $[0,1]$  with 1000 points and then stack the results into a tall array with rows being the $(cos(x),sin(x))$ entries.

```python
# Write your code here

x = np.linspace(0,1,1000)

cos_vals = np.cos(x)
sin_vals = np.sin(x)

result = np.column_stack((cos_vals,sin_vals))

print(result)

print("Shape: ",result.shape)
```

# Problem: 9

Create a random $3×5$ array using the `np.random.rand(3 ,5)` function and compute: the sum of all the entries, the sum of the rows, and the sum of the columns.

```python
# Write your code here

arr = np.random.rand(3 ,5) # this creates 2d array with shape 3 x 5 and with random numbers between 0 and 1

total_sum = arr.sum()
row_sum = arr.sum(axis=1)
col_sum = arr.sum(axis=0)

print("Array:\n", arr)
print("\nSum of all the entries: ",total_sum)
print("Sum of Rows: ",row_sum)
print("Sum of the columns: ",col_sum)
```
